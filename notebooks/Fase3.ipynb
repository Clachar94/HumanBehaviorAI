## 2. Limpieza de Datos

En esta sección se identifican y corrigen los problemas de calidad presentes en el dataset original,
incluyendo valores nulos, duplicados, formatos incorrectos y consistencia general de los datos.

### 2.1 Detección de Valores Nulos

```python
print(" ANÁLISIS DE VALORES NULOS")
print("=" * 50)

nulos = df_raw.isnull().sum()
pct_nulos = (nulos / len(df_raw) * 100).round(2)

reporte_nulos = pd.DataFrame({
    'Valores nulos': nulos,
    'Porcentaje (%)': pct_nulos,
    'Tipo de dato': df_raw.dtypes
})

reporte_nulos['¿Tiene nulos?'] = reporte_nulos['Valores nulos'].apply(
    lambda x: ' SÍ' if x > 0 else ' NO'
)

display(reporte_nulos)
print(f"\n Total de columnas con nulos: {(nulos > 0).sum()}")
print(f" Total de valores nulos en el dataset: {nulos.sum():,}")
```

### Interpretación
El análisis de nulos muestra que solo la columna `Hashtags` tiene valores faltantes, 
con 8,341 registros que representan el 16.68% del dataset. Todas las demás columnas 
están completas, incluyendo las más críticas como `User ID`, `Bot Label` y `Created At`. 
Esto es una buena señal ya que los nulos están concentrados en una sola columna secundaria 
y no afectan la integridad general del dataset.

### Interpretación
El gráfico confirma visualmente que únicamente la columna `Hashtags` presenta valores 
nulos en todo el dataset. Con 8,341 registros sin hashtag, la decisión de eliminar esas 
filas no es viable ya que se perdería el 16.68% de los datos. Por eso se optará por 
reemplazar los nulos con el valor "sin_hashtags", conservando todos los registros para 
el análisis y el entrenamiento de los modelos.

### 2.2 Tratamiento de Valores Nulos

La columna `Hashtags` presenta **8,341 valores nulos (16.68%)**.

**Decisión tomada:** Se reemplazan por la cadena `"sin_hashtags"`.

**Justificación:** La ausencia de hashtags en un tweet es en sí misma información relevante
para la detección de bots. Eliminar esas filas implicaría perder el 16.68% del dataset,
lo cual afectaría negativamente el entrenamiento del modelo. Reemplazar con un valor
neutro identificable conserva la información de que no se usaron hashtags.

```python
# Trabajamos sobre una copia para preservar el original
df = df_raw.copy()

# Tratamiento de nulos en Hashtags
df['Hashtags'] = df['Hashtags'].fillna('sin_hashtags')

# Verificación
print("TRATAMIENTO DE NULOS COMPLETADO")
print("=" * 50)
print(f"Nulos en 'Hashtags' antes: {df_raw['Hashtags'].isnull().sum():,}")
print(f"Nulos en 'Hashtags' después: {df['Hashtags'].isnull().sum()}")
print(f"\n Total de nulos restantes en el dataset: {df.isnull().sum().sum()}")
```

### 2.3 Detección y Eliminación de Duplicados

```python
duplicados = df.duplicated().sum()
duplicados_user = df.duplicated(subset=['User ID']).sum()

print(" ANÁLISIS DE DUPLICADOS")
print("=" * 50)
print(f"Filas completamente duplicadas:     {duplicados:,}")
print(f"User IDs duplicados:                {duplicados_user:,}")

if duplicados > 0:
    df = df.drop_duplicates()
    print(f"\n Se eliminaron {duplicados:,} filas duplicadas.")
else:
    print("\n No se encontraron filas duplicadas.")

print(f"\n Tamaño del dataset tras limpieza: {df.shape[0]:,} filas × {df.shape[1]} columnas")
```

### Interpretación
El análisis de duplicados confirma que el dataset no tiene filas repetidas ni 
User IDs duplicados, lo que indica que cada registro corresponde a un usuario 
único. Se conservan los 50,000 registros completos para las siguientes etapas.

### 2.4 Corrección de Tipos de Datos y Formatos

```python
print(" CORRECCIÓN DE TIPOS DE DATOS")
print("=" * 50)

# 1. Convertir Created At a datetime
df['Created At'] = pd.to_datetime(df['Created At'])
print(" 'Created At' convertido a datetime")

# 2. Extraer componentes de fecha útiles
df['anio_creacion']  = df['Created At'].dt.year
df['mes_creacion']   = df['Created At'].dt.month
df['hora_creacion']  = df['Created At'].dt.hour
df['dia_semana']     = df['Created At'].dt.dayofweek  # 0=Lunes, 6=Domingo
print(" Columnas de fecha extraídas: año, mes, hora, día de semana")

# 3. Estandarizar texto en columnas de cadena
df['Location'] = df['Location'].str.strip().str.title()
df['Hashtags'] = df['Hashtags'].str.strip().str.lower()
print(" Columnas de texto estandarizadas (strip + capitalización)")

# 4. Verificar tipos finales
print("\n Tipos de datos actuales:")
print(df.dtypes)
```

### Interpretación
Se realizaron tres correcciones importantes. Primero, la columna `Created At` se convirtió 
de texto a datetime, lo que permite extraer componentes de fecha como año, mes, hora y día 
de la semana que serán útiles para detectar patrones temporales de los bots. Segundo, se 
estandarizó el texto de `Location` y `Hashtags` para evitar inconsistencias como mayúsculas 
o espacios extra. El dataset ahora tiene 15 columnas con tipos de datos correctos y listos 
para el procesamiento.

## 3. Transformaciones y Creación de Variables de Comportamiento

En esta etapa se crean nuevas variables a partir de los datos limpios. Estas variables
representan numéricamente patrones de comportamiento que los modelos de IA utilizarán
para distinguir bots de usuarios humanos.

### 3.1 Variables de Actividad y Engagement

```python
print(" CREANDO VARIABLES DE ACTIVIDAD Y ENGAGEMENT")
print("=" * 55)

# 1. Ratio de engagement: interacciones totales por follower
df['engagement_ratio'] = (
    (df['Retweet Count'] + df['Mention Count']) /
    (df['Follower Count'] + 1)  # +1 para evitar división por cero
).round(4)

# 2. Actividad total: suma de retweets y menciones
df['actividad_total'] = df['Retweet Count'] + df['Mention Count']

# 3. Ratio retweet vs mención
df['ratio_retweet_mencion'] = (
    df['Retweet Count'] / (df['Mention Count'] + 1)
).round(4)

# 4. Cuenta verificada como entero (para modelos)
df['verificado_int'] = df['Verified'].astype(int)

print(" Variables creadas:")
print("   - engagement_ratio        → interacciones / seguidores")
print("   - actividad_total         → retweets + menciones")
print("   - ratio_retweet_mencion   → retweets / menciones")
print("   - verificado_int          → Verified como 0/1")

display(df[['User ID', 'engagement_ratio', 'actividad_total',
            'ratio_retweet_mencion', 'verificado_int']].head())
```

### Interpretación
Se crearon 4 variables de actividad. El `engagement_ratio` mide qué tan activa es una 
cuenta en relación a sus seguidores — un bot tiende a tener un ratio muy alto o muy bajo 
comparado con usuarios normales. La `actividad_total` suma retweets y menciones para tener 
una métrica general de interacción. El `ratio_retweet_mencion` indica si una cuenta retweetea 
mucho más de lo que menciona, comportamiento típico de bots que amplifican contenido 
automáticamente. Finalmente `verificado_int` convierte el booleano a número para que los 
modelos puedan procesarlo.

### 3.2 Variables Temporales y de Horario

```python
print(" CREANDO VARIABLES TEMPORALES")
print("=" * 55)

# 1. Franja horaria del día
def franja_horaria(hora):
    if 0 <= hora < 6:
        return 'madrugada'
    elif 6 <= hora < 12:
        return 'mañana'
    elif 12 <= hora < 18:
        return 'tarde'
    else:
        return 'noche'

df['franja_horaria'] = df['hora_creacion'].apply(franja_horaria)

# 2. Es fin de semana
df['es_fin_de_semana'] = df['dia_semana'].apply(
    lambda x: 1 if x >= 5 else 0
)

# 3. Antigüedad de la cuenta en días (respecto al registro más reciente)
fecha_referencia = df['Created At'].max()
df['antiguedad_dias'] = (fecha_referencia - df['Created At']).dt.days

print(" Variables creadas:")
print("   - franja_horaria    → madrugada / mañana / tarde / noche")
print("   - es_fin_de_semana  → 1 si sábado o domingo, 0 si no")
print("   - antiguedad_dias   → días desde creación hasta la cuenta más reciente")

display(df[['User ID', 'hora_creacion', 'franja_horaria',
            'es_fin_de_semana', 'antiguedad_dias']].head())
```

### Interpretación
Las variables temporales permiten detectar patrones de comportamiento relacionados con 
el horario. Los bots suelen estar activos en horas poco comunes como la madrugada ya 
que operan de forma automática sin importar la hora. La `franja_horaria` agrupa la 
actividad en cuatro bloques del día, `es_fin_de_semana` indica si la cuenta estuvo 
activa en fin de semana y `antiguedad_dias` mide qué tan nueva es la cuenta, ya que 
las cuentas bot suelen ser más recientes que las de usuarios humanos reales.

### 3.3 Variables de Contenido (Hashtags y Tweets)

```python
print("CREANDO VARIABLES DE CONTENIDO")
print("=" * 55)

# 1. Cantidad de hashtags usados
df['cantidad_hashtags'] = df['Hashtags'].apply(
    lambda x: 0 if x == 'sin_hashtags' else len(str(x).split())
)

# 2. Largo del tweet en caracteres
df['largo_tweet'] = df['Tweet'].astype(str).apply(len)

# 3. Cantidad de palabras en el tweet
df['palabras_tweet'] = df['Tweet'].astype(str).apply(
    lambda x: len(x.split())
)

# 4. Usa hashtags: flag binaria
df['usa_hashtags'] = df['cantidad_hashtags'].apply(
    lambda x: 1 if x > 0 else 0
)

print(" Variables creadas:")
print("   - cantidad_hashtags  → número de hashtags por tweet")
print("   - largo_tweet        → caracteres totales del tweet")
print("   - palabras_tweet     → palabras totales del tweet")
print("   - usa_hashtags       → 1 si usa hashtags, 0 si no")

display(df[['User ID', 'Tweet', 'cantidad_hashtags',
            'largo_tweet', 'palabras_tweet', 'usa_hashtags']].head())
```

### Interpretación
Las variables de contenido analizan el comportamiento de los tweets publicados. Los bots 
tienden a publicar tweets con patrones similares en cuanto a largo y cantidad de palabras, 
ya que el contenido es generado automáticamente. La `cantidad_hashtags` y `usa_hashtags` 
son importantes porque algunos bots usan hashtags masivamente para posicionar tendencias, 
mientras que otros no usan ninguno. El `largo_tweet` y `palabras_tweet` ayudan a detectar 
tweets generados automáticamente que suelen tener longitudes muy consistentes entre sí.

### 3.4 Variables de Riesgo de Bot

```python
print(" CREANDO VARIABLES DE RIESGO")
print("=" * 55)

# 1. Pocos seguidores con alta actividad (patrón típico de bot)
df['alta_actividad_pocos_seguidores'] = (
    (df['Follower Count'] < df['Follower Count'].quantile(0.25)) &
    (df['actividad_total'] > df['actividad_total'].quantile(0.75))
).astype(int)

# 2. Cuenta no verificada con engagement muy alto
df['no_verif_alto_engagement'] = (
    (df['verificado_int'] == 0) &
    (df['engagement_ratio'] > df['engagement_ratio'].quantile(0.90))
).astype(int)

# 3. Actividad en madrugada (comportamiento más común en bots)
df['activo_madrugada'] = (df['franja_horaria'] == 'madrugada').astype(int)

# 4. Score de sospecha simple (suma de indicadores de riesgo)
df['score_sospecha'] = (
    df['alta_actividad_pocos_seguidores'] +
    df['no_verif_alto_engagement'] +
    df['activo_madrugada']
)

print(" Variables creadas:")
print("   - alta_actividad_pocos_seguidores  → patrón de bot clásico")
print("   - no_verif_alto_engagement         → no verificado + engagement alto")
print("   - activo_madrugada                 → actividad entre 00:00 y 06:00")
print("   - score_sospecha                   → suma de indicadores (0 a 3)")

display(df[['User ID', 'Bot Label', 'alta_actividad_pocos_seguidores',
            'no_verif_alto_engagement', 'activo_madrugada',
            'score_sospecha']].head(8))
```

### Interpretación
Las variables de riesgo combinan múltiples indicadores para generar señales de alerta. 
`alta_actividad_pocos_seguidores` detecta cuentas que interactúan mucho pero tienen pocos 
seguidores, lo cual es un patrón clásico de bot. `no_verif_alto_engagement` identifica 
cuentas no verificadas con un engagement desproporcionalmente alto. `activo_madrugada` 
marca las cuentas que publican entre las 00:00 y las 06:00, horario poco común para 
usuarios humanos reales. El `score_sospecha` suma estos tres indicadores generando un 
puntaje de 0 a 3, donde 3 indica el mayor nivel de sospecha. En los primeros registros 
se puede ver que varios bots tienen score 1, lo que confirma que al menos uno de los 
patrones de riesgo se cumple.

la verificación tampoco es un indicador definitivo por sí solo. Esto refuerza la necesidad

## 7. Carga a SQLite

Se cargan los datos en una base de datos SQLite (`humanbehaviorai.db`), en dos tablas: `raw_events` (los datos originales, tal como llegaron) y `processed_sessions` (los datos ya limpios, con las variables de comportamiento calculadas). Esto permite consultar la información con SQL en la siguiente sección.

```python
import os
import sqlite3
from datetime import datetime

# Conexión a la base de datos
ruta_db = os.path.join("..", "data", "processed", "humanbehaviorai.db")
conn = sqlite3.connect(ruta_db)

# Cargar tabla raw_events (datos originales)
df_raw_sql = df_raw.rename(columns={
    'User ID': 'user_id', 'Username': 'username', 'Tweet': 'tweet',
    'Retweet Count': 'retweet_count', 'Mention Count': 'mention_count',
    'Follower Count': 'follower_count', 'Verified': 'verified',
    'Bot Label': 'bot_label', 'Location': 'location',
    'Created At': 'created_at', 'Hashtags': 'hashtags'
})
df_raw_sql.to_sql('raw_events', conn, if_exists='replace', index=False)
print(f"Tabla raw_events cargada: {len(df_raw_sql):,} registros")

# Cargar tabla processed_sessions (datos procesados)
columnas = ['User ID', 'engagement_ratio', 'actividad_total', 'ratio_retweet_mencion',
            'franja_horaria', 'es_fin_de_semana', 'antiguedad_dias',
            'cantidad_hashtags', 'largo_tweet', 'palabras_tweet',
            'usa_hashtags', 'score_sospecha', 'Bot Label']
df_proc_sql = df[columnas].copy()
df_proc_sql.columns = ['user_id', 'engagement_ratio', 'actividad_total', 'ratio_retweet_mencion',
                        'franja_horaria', 'es_fin_de_semana', 'antiguedad_dias',
                        'cantidad_hashtags', 'largo_tweet', 'palabras_tweet',
                        'usa_hashtags', 'score_sospecha', 'bot_label']
df_proc_sql['processed_at'] = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
df_proc_sql.to_sql('processed_sessions', conn, if_exists='replace', index=False)
print(f"Tabla processed_sessions cargada: {len(df_proc_sql):,} registros")

conn.close()
print("Carga a SQLite completada")
```

## 8. Consultas a la Base de Datos

Se ejecutan algunas consultas SQL sobre la tabla `processed_sessions`, para demostrar que los datos quedaron correctamente cargados y son consultables.

```python
conn = sqlite3.connect(ruta_db)

print("=== Consulta 1: Conteo de cuentas por tipo (bot vs humano) ===")
q1 = "SELECT bot_label, COUNT(*) AS cantidad FROM processed_sessions GROUP BY bot_label"
display(pd.read_sql_query(q1, conn))

print("\n=== Consulta 2: Promedio de engagement_ratio por tipo de cuenta ===")
q2 = "SELECT bot_label, ROUND(AVG(engagement_ratio), 4) AS engagement_promedio FROM processed_sessions GROUP BY bot_label"
display(pd.read_sql_query(q2, conn))

print("\n=== Consulta 3: Top 5 cuentas con mayor score de sospecha ===")
q3 = "SELECT user_id, score_sospecha, engagement_ratio, franja_horaria FROM processed_sessions ORDER BY score_sospecha DESC LIMIT 5"
display(pd.read_sql_query(q3, conn))

print("\n=== Consulta 4: Cuentas activas de madrugada, por tipo ===")
q4 = "SELECT bot_label, COUNT(*) AS cantidad FROM processed_sessions WHERE franja_horaria = 'madrugada' GROUP BY bot_label"
display(pd.read_sql_query(q4, conn))

conn.close()
```
